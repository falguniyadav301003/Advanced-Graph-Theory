```python
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

# Node class

class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None


# Create tree from user input

def create_tree():
    data = input("Enter root value: ")

    if data == "-1":
        return None

    root = Node(int(data))
    queue = [root]

    # Keep track of values already used
    used = {root.data}

    while queue:
        current = queue.pop(0)

        # Left child
        while True:
            left = input(
                f"Enter left child of {current.data} (-1 for no node): "
            )

            if left == "-1":
                break

            left = int(left)

            if left in used:
                print("Error: This value already exists in the tree.")
                print("Enter -1 or a new value.")
            else:
                current.left = Node(left)
                used.add(left)
                queue.append(current.left)
                break

        # Right child
        while True:
            right = input(
                f"Enter right child of {current.data} (-1 for no node): "
            )

            if right == "-1":
                break

            right = int(right)

            if right in used:
                print("Error: This value already exists in the tree.")
                print("Enter -1 or a new value.")
            else:
                current.right = Node(right)
                used.add(right)
                queue.append(current.right)
                break

    return root


# Preorder
# Root → Left → Right

def preorder(root, result):
    if root:
        result.append(root)
        preorder(root.left, result)
        preorder(root.right, result)



# Inorder
# Left → Root → Right

def inorder(root, result):
    if root:
        inorder(root.left, result)
        result.append(root)
        inorder(root.right, result)



# Postorder
# Left → Right → Root

def postorder(root, result):
    if root:
        postorder(root.left, result)
        postorder(root.right, result)
        result.append(root)



# Get positions of nodes

def get_positions(root):

    positions = {}

    def assign(node, x, y, level):

        if node is None:
            return

        positions[node] = (x, y)

        # Increase horizontal separation at lower levels
        distance = 3 / (2 ** level)

        assign(
            node.left,
            x - distance,
            y - 1,
            level + 1
        )

        assign(
            node.right,
            x + distance,
            y - 1,
            level + 1
        )

    assign(root, 0, 0, 0)

    return positions



# Animate traversal

def animate_traversal(traversal, title):

    fig, ax = plt.subplots(figsize=(12, 7))

    positions = get_positions(root)

    def update(frame):

        ax.clear()


        # DRAW EDGES

        for node, (x, y) in positions.items():

            if node.left:

                x2, y2 = positions[node.left]

                ax.plot(
                    [x, x2],
                    [y, y2],
                    color="black",
                    linewidth=2
                )

            if node.right:

                x2, y2 = positions[node.right]

                ax.plot(
                    [x, x2],
                    [y, y2],
                    color="black",
                    linewidth=2
                )



        # DRAW NODES

        for node, (x, y) in positions.items():

            # Default colour
            node_color = "lightblue"

            # Already visited → GREEN
            if node in traversal[:frame]:
                node_color = "lightgreen"

            # Current node → YELLOW
            if (
                frame < len(traversal)
                and node == traversal[frame]
            ):
                node_color = "yellow"

            ax.scatter(
                x,
                y,
                s=1800,
                color=node_color,
                edgecolor="black",
                linewidth=2,
                zorder=2
            )

            ax.text(
                x,
                y,
                str(node.data),
                ha="center",
                va="center",
                fontsize=16,
                fontweight="bold",
                zorder=3
            )


        # TRAVERSAL INFORMATION

        sequence = [
            str(node.data)
            for node in traversal
        ]

        visited = [
            str(node.data)
            for node in traversal[:frame]
        ]

        if frame < len(traversal):

            current = str(
                traversal[frame].data
            )

        else:

            current = "Done"


        ax.set_title(
            f"{title}\n\n"
            f"Traversal Order: {' → '.join(sequence)}\n"
            f"Visited: "
            f"{' → '.join(visited) if visited else 'None'}\n"
            f"Current Node: {current}",
            fontsize=15,
            fontweight="bold"
        )

        ax.axis("off")

        # Keep graph centered
        if positions:

            xs = [p[0] for p in positions.values()]
            ys = [p[1] for p in positions.values()]

            ax.set_xlim(
                min(xs) - 1,
                max(xs) + 1
            )

            ax.set_ylim(
                min(ys) - 1,
                max(ys) + 1
            )


    # CREATE ANIMATION

    animation = FuncAnimation(
        fig,
        update,
        frames=len(traversal) + 1,
        interval=1200,
        repeat=False
    )

    plt.close(fig)

    display(
        HTML(
            animation.to_jshtml()
        )
    )


# MAIN PROGRAM

root = create_tree()


# Generate traversals

pre = []
ino = []
post = []

preorder(root, pre)
inorder(root, ino)
postorder(root, post)


# Print results

print("TRAVERSAL RESULTS")

print(
    "Preorder:",
    [node.data for node in pre]
)

print(
    "Inorder:",
    [node.data for node in ino]
)

print(
    "Postorder:",
    [node.data for node in post]
)


# Show animations

print("\n▶ PREORDER ANIMATION")
animate_traversal(
    pre,
    "PREORDER: Root → Left → Right"
)


print("\n▶ INORDER ANIMATION")
animate_traversal(
    ino,
    "INORDER: Left → Root → Right"
)


print("\n▶ POSTORDER ANIMATION")
animate_traversal(
    post,
    "POSTORDER: Left → Right → Root"
)
```

```python
# SHORTEST PATH ANIMATION

from collections import deque
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display


# FIND NODE

def find_node(root, value):

    if root is None:
        return None

    queue = deque([root])

    while queue:

        node = queue.popleft()

        if node.data == value:
            return node

        if node.left:
            queue.append(node.left)

        if node.right:
            queue.append(node.right)

    return None


# FIND SHORTEST PATH

def shortest_path(root, source, destination):

    source_node = find_node(root, source)
    destination_node = find_node(root, destination)

    if source_node is None or destination_node is None:
        return []

    # Store parent of every node
    parent = {root: None}

    queue = deque([root])

    while queue:

        current = queue.popleft()

        if current.left:
            parent[current.left] = current
            queue.append(current.left)

        if current.right:
            parent[current.right] = current
            queue.append(current.right)

    # Source → Root
    source_path = []

    current = source_node

    while current is not None:
        source_path.append(current)
        current = parent[current]

    # Destination → Root
    destination_path = []

    current = destination_node

    while current is not None:
        destination_path.append(current)
        current = parent[current]

    # Find common ancestor
    source_set = set(source_path)

    common = None

    for node in destination_path:

        if node in source_set:
            common = node
            break

    # Source → Common ancestor
    path = []

    current = source_node

    while current != common:

        path.append(current)
        current = parent[current]

    path.append(common)

    # Common ancestor → Destination
    second_part = []

    current = destination_node

    while current != common:

        second_part.append(current)
        current = parent[current]

    second_part.reverse()

    path.extend(second_part)

    return path


# SHORTEST PATH ANIMATION


def shortest_path_animation(root, path, source, destination):

    positions = get_positions(root)

    fig, ax = plt.subplots(figsize=(12, 7))


    def update(frame):

        ax.clear()


        # DRAW ALL TREE EDGES


        for node, (x, y) in positions.items():

            if node.left:

                x2, y2 = positions[node.left]

                ax.plot(
                    [x, x2],
                    [y, y2],
                    color="black",
                    linewidth=2
                )

            if node.right:

                x2, y2 = positions[node.right]

                ax.plot(
                    [x, x2],
                    [y, y2],
                    color="black",
                    linewidth=2
                )


        # HIGHLIGHT PATH UP TO CURRENT STEP

        current_path = path[:frame + 1]

        # Highlight path edges
        for i in range(len(current_path) - 1):

            node1 = current_path[i]
            node2 = current_path[i + 1]

            x1, y1 = positions[node1]
            x2, y2 = positions[node2]

            ax.plot(
                [x1, x2],
                [y1, y2],
                color="red",
                linewidth=6,
                zorder=1
            )

        # DRAW NODES

        for node, (x, y) in positions.items():

            # Normal node
            node_color = "lightblue"

            # Nodes already included in path
            if node in current_path:

                node_color = "lightgreen"

            # Current node
            if (
                frame < len(path)
                and node == path[frame]
            ):

                node_color = "yellow"

            # Source node
            if node == path[0]:

                node_color = "orange"

            # Destination when reached
            if (
                node == path[-1]
                and frame == len(path) - 1
            ):

                node_color = "limegreen"


            ax.scatter(
                x,
                y,
                s=1800,
                color=node_color,
                edgecolor="black",
                linewidth=2,
                zorder=2
            )

            ax.text(
                x,
                y,
                str(node.data),
                ha="center",
                va="center",
                fontsize=16,
                fontweight="bold",
                zorder=3
            )

        # INFORMATION

        reached = [
            str(node.data)
            for node in current_path
        ]

        full_path = [
            str(node.data)
            for node in path
        ]


        if frame < len(path):

            current = path[frame].data

        else:

            current = "Completed"


        ax.set_title(
            f"SHORTEST PATH ANIMATION\n\n"
            f"Source: {source}    "
            f"Destination: {destination}\n\n"
            f"Shortest Path: "
            f"{' → '.join(full_path)}\n\n"
            f"Current Path: "
            f"{' → '.join(reached)}\n\n"
            f"Current Node: {current}",
            fontsize=14,
            fontweight="bold"
        )

        ax.axis("off")


        # Keep graph centered

        xs = [
            p[0]
            for p in positions.values()
        ]

        ys = [
            p[1]
            for p in positions.values()
        ]

        ax.set_xlim(
            min(xs) - 1,
            max(xs) + 1
        )

        ax.set_ylim(
            min(ys) - 1,
            max(ys) + 1
        )


    # CREATE ANIMATION

    animation = FuncAnimation(
        fig,
        update,
        frames=len(path),
        interval=1200,
        repeat=False
    )

    plt.close(fig)

    display(
        HTML(
            animation.to_jshtml()
        )
    )


# TAKE SOURCE AND DESTINATION

print("SHORTEST PATH")


source = int(
    input("Enter source node: ")
)

destination = int(
    input("Enter destination node: ")
)


# FIND PATH

path = shortest_path(
    root,
    source,
    destination
)


# DISPLAY RESULT + ANIMATION

if not path:

    print(
        "\nSource or destination node does not exist."
    )

else:

    print("\nShortest Path:")

    print(
        " → ".join(
            str(node.data)
            for node in path
        )
    )

    print(
        "Number of edges:",
        len(path) - 1
    )

    print(
        "\n▶ Starting shortest path animation..."
    )

    shortest_path_animation(
        root,
        path,
        source,
        destination
    )
```
